```python
class Solution:
    def detectCycle(self, head):
        slow = fast = head

        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next

            if slow == fast:
                break
        else:
            return None

        ptr = head

        while ptr != slow:
            ptr = ptr.next
            slow = slow.next

        return ptr
```
