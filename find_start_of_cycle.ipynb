```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def detect_cycle_start(head):
    slow = head
    fast = head

    # Step 1: detect cycle
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next

        if slow == fast:
            # Step 2: find cycle start
            slow = head

            while slow != fast:
                slow = slow.next
                fast = fast.next

            return slow  # start of cycle

    return None  # no cycle
```
