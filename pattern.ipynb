```python
n = 5

for i in range(n):
    for j in range(n):
        if i == 0 or i == n-1 or j == 0 or j == n-1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
n = 5

for i in range(1,n+1):
    print(" "*(n-i),end="")
    for j in range(1,2*i):
        if j==1 or j==2*i-1:
            print("*",end="")
        else:
            print(" ",end="")
    print()

for i in range(n-1,0,-1):
    print(" "*(n-i),end="")
    for j in range(1,2*i):
        if j==1 or j==2*i-1:
            print("*",end="")
        else:
            print(" ",end="")
    print()
```
