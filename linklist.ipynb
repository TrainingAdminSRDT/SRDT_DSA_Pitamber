```python
# Linked List Implementation in Python

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

# Create nodes
head = Node(10)
head.next = Node(20)
head.next.next = Node(30)

# Traverse and print
current = head
while current:
    print(current.data, end=" -> ")
    current = current.next
print("None")
```
